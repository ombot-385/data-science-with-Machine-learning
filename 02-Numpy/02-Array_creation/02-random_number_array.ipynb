we can create random number arrays from a variety of distributions using
NumPy functions and methods (great for sampling and simulation!)

```python
from numpy.random import default_rng
rng=default_rng(12345)
random_array=rng.random(10)
random_array
```

```python
rng=default_rng(12345)
mean, stddev=5,1
random_normal=rng.normal(mean,stddev,size=10)
random_normal
```
