As an alternative to converting lists, we can create arrays using functions

#### ones

```python
## ones
import numpy as np
print(np.ones(4,))
```

```python
np.ones((3,4))
```

```python
np.ones((3,4),dtype=bool)
```

```python
np.ones((3,4),dtype=float)
```

#### zeros

```python
np.zeros((3,4),dtype=int)
```

```python
np.zeros((3,5),dtype=float)
```

#### arange

```python
np.arange(1,11,2)
```

```python
np.arange(1,11,3)
```

#### linspace

```python
np.linspace(0,100,10)
```

```python
np.linspace(1,10,100)
```

#### reshape

```python
cell=np.arange(0,18,2)
cell
```

```python
cell.reshape(3,3)
```
