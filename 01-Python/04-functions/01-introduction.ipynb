```python
def even_or_odd(num):
    """This function finds even or odd"""
    if(num%2==0):
        print("the number is even")
    else:
        print("the number is odd") 

## calling the function
even_or_odd(24)
```

```python
## Default parameters
def greet(name):
    print(f"hello hello {name}")

greet("omkar")

def greet(name="guest"):
    print(f"hello {name}")
greet()
greet("omkar")
```

```python
## variable length argument
## Positional and keywords arguments

def print_numbers(*args): ## instead of args any variable like omkar can be used
    for number in args:
        print(number)

print_numbers(1,2,3,4,5,6,7,8,"omkar")
```

```python
## keywords arguments
def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")

print_details(name="omkar",age="32",country="india")
```

```python
## keywords arguments
def print_details(*args,**kwargs):
    for val in args:
        print(f"positional arguments:{val}")

    for key,value in kwargs.items():
        print(f"{key}:{value}")

print_details(name="omkar",age="32",country="india")
print_details(1,2,3,4,"omkar",name="omkar",age="32",country="india")
```
