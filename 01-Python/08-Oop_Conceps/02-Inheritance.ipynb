Inheritance is a fundamental concept in Object-Oriented Programming (OOP) that allows a class to inherit attributes and methods from another class.

```python
class car:
    def __init__(self,windows,doors,enginetype):
        self.windows=windows
        self.doors=doors
        self.enginetype=enginetype

    def drive(self):
        print(f"The person will drive the {self.enginetype} car")
```

```python
car1=car(4,5,"petrol")
car1.drive()
```

```python
class Tesla(car):
    def __init__(self,windows,doors,enginetype,is_selfdriving):
        '''The super() call: This line literally jumps back up to the Car class and runs its code first.'''
        '''super() means go to the parent class and use its method'''
        super().__init__(windows,doors,enginetype)
        self.is_selfdriving=is_selfdriving

    def selfdriving(self):
        print(f"Tesla supports self driving:{self.is_selfdriving}")
```

```python
tesla1=Tesla(4,5,"electric",True)
tesla1.selfdriving()
```

```python
tesla1.drive()
```

```python
### Multiple Inheritance
## When a class inherits from more than one base class.

## base class 1
class Animal:
    def __init__(self,name):
        self.name=name

    def speak():
        print("subclass must implement this method")

## Base class 2
class Pet:
    def __init__(self,owner):
            self.owner=owner

## derived class
class Dog(Animal,Pet):
    def __init__(self,name,owner):
          Animal.__init__(self,name)
          Pet.__init__(self,owner)

    def speak(self):
         return f"{self.name} says woof"

## create an object
dog=Dog("Buddy","omkar")
print(dog.speak())
print(f"owner:{dog.owner}")
```
