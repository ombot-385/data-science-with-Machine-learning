dictionaries are unordered collection of items. they store data in key value pairs .keys must be unique and immutable (eg. strings,numbers,or tuples), while values can be of any type

```python
## creating dictionaries
empty_dictionaries={}
print(type(empty_dictionaries))
```

```python
empty_dict=dict()
empty_dict
```

```python
student={"name":"krish","age":32,"grade":24}
print(student)
print(type(student))
```

```python
## single key is always used
student={"name":"krish","age":32,"name":24}
print(student)
print(type(student))
```

```python
## accesing dictionary elements
student={"name":"krish","age":32,"grade":'A'}
print(student['name'])
print(student['age'])
print(student['grade'])


## accesing using get() method
print(student.get('name'))
print(student.get('age'))
print(student.get('grade'))
print(student.get('last_name'))
```

```python
## modifying dictionary elements
## Dictionary are mutuble , so you can add,update or delete elements

print(student)
student["age"]=25 ## update value for the key
print(student)

student["address"]="india"  ## added the new key
print(student)
```

```python
## deleting the key
del student['grade']
print(student)
```

```python
## dictionary methods
keys=student.keys() ## get all the keys
print(keys)

values=student.values()  ## get all values
print(values)

items=student.items() ## get all key value pairs
print(items)
```

```python
## shallow copy
student_copy=student
print(student)
print(student_copy)
```

```python
student["name"]="om"
print(student)
print(student_copy)
```

```python
student_copy1=student.copy()  ## shallow copy
print(student)
print(student_copy1)
```

```python
student["name"]="omkar"
print(student)
print(student_copy1)
```

```python
## iterating over dictionaries
## we can use for loops to iterate over dictionaries,keys,values or items

## iterating over keys
for keys in student.keys():
    print(keys)
```

```python
## iterate over values
for value in student.values():
    print(value)
```

```python
## itearate over key value pairs
for key,value in student.items():
    print(f"{key}:{value}")
```

```python
## Nested dictionaries
students={
    "student1":{"name":"omkar","age":20},
    "student2":{"name":"spiderman","age":35}

}

print(students)
```

```python
## access nested dictionaries elements
print(students["student2"]["name"])
print(students["student1"]["age"])
```

```python
students.items()
```

```python
## iterating over nested dictionaries
for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")

    
```

```python
## iterating over nested dictionaries
for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")

    for key,value in student_info.items():
        print(f"{key}:{value}")

    
```

```python
## dictionary comprehension
squares={x:x**2 for x in range(5)}
print(squares)
```

```python
## conditional dictionaries comprehension
evens={x:x**2 for x in range (10) if x%2==0}
print(evens)
```

```python
## practical examples
## use a dictionary to count the frequency of elements in list

numbers=[1,2,2,3,3,3,4,4,4,4]
frequency={}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```

```python
## merge two dictionary into one
dict1={"a":1,"b":2}
dict2={"b":3,"c":4}

merged_dict={**dict1,**dict2}
print(merged_dict)
```
