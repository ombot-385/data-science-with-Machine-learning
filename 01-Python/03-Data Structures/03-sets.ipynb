sets are abuilt-in data type in python used to store collections of unique items. they are unordered, meaning that the elements do not follow a specific order, and they do not allow duplicates.sets are useful memebership tests, eliminating duplicate entries,and performing mathematical set operations like union,intersection,differenec, and symmetric difference.

```python
## creating a set
my_set={1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set=set()
print(type(my_empty_set))
```

```python
my_set2=set([1,2,3,4,5,5])
print(my_set2)
print(type(my_set2))
```

```python
## Basic set operations
## adding elements

my_set2.add(7)
print(my_set2)

my_set2.add(7)
print(my_set2)
```

```python
## Remove the elemets from set
my_set2.remove(3)
print(my_set2)
```

```python
my_set.discard(11) ## discard-->remove an element from a set if it is a member
```

```python
## pop methods
removed_element=my_set2.pop()
print(removed_element)
print(my_set2)
```

```python
## claer all the elements
my_set2.clear()
print(my_set2)
```

```python
## set membership test
my_set={1,2,3,4,5}
print(3 in my_set)
print(10 in my_set)
```

```python
## mathematical test
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

## union
union_set=set1.union(set2)
print(union_set)

##intersection set
intersection_set=set1.intersection(set2)
print(intersection_set)

set1.intersection_update(set2)
print(set1)
```

```python
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

## difference
print(set1.difference(set2))
print(set2.difference(set1))
```

```python
## symmetric difference --> takes unique elements from both set
set1.symmetric_difference(set2)
```

```python
## sets method
set1={1,2,3,4,5}
set2={3,4,5}

## is subset
print(set1.issubset(set2))
print(set1.issuperset(set2))
```

```python
lst=[1,2,2,3,4,4,5]
print(set(lst))
set(lst)
```

```python
## counting unique words in test
text="in this file we are discussing about sets"
words=text.split()
print(words)

## convert list of words to set to get unique words
unique_words=set(words)
print(unique_words)
print(len(unique_words))
```
