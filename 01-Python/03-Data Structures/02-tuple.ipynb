```python
## Accesing tuple elements
print(number)
print(number[2])
print(number[-1])
print(number[0:4])
```

```python
## Tuples operation
concatenation_tuple=number+mixed_tuple
print(concatenation_tuple)

print(concatenation_tuple*3)
concatenation_tuple*3
```

```python
## Immutable nature of Tuples

print("Lists are mutable")
lst=[1,2,3,4,5]
print(lst)

lst[1]="om"
print(lst)

print("Tuples are immutable")

number=(1,2,3,4,5)
## number[1]=5 --> we cant do this as tuple are immutable
```

```python
## Tuple methods
print(number.count(1))
print(number.index(3))
```

```python
## Packing and Unpacking tuple

#Packing
packed_tuple=1,"hello",3.14
print(packed_tuple)

#Unpacking a tuple
a,b,c=packed_tuple

print(a)
print(b)
print(c)
```

```python
## Unpacking with *
numbers=(1,2,3,4,5,6)
first,*middle,last=numbers
print(first)
print(middle)
print(last)
```

```python
## Nested list
lst=[[1,2,3,4],[6,7,8,9],[1,"hello",3.14,"c"]]
print(lst[0])
print(lst[0][2])
print(lst[0][0:3])

## 
```

```python
## Nested tuple
nested_tuple=((1,2,3),("a","b","c"),(True,False))
print(nested_tuple[0])
print(nested_tuple[1],[2])
print(nested_tuple[1][2])
```

```python
## Iterating over nested tuples
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end=" ")
    print()
```
