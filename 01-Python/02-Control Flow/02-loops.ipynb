## For loop

```python
## for loop
for i in range(5):
    print(i)
```

```python
for i in range(1,6):
    print(i)
```

```python
for i in range(1,6,2):
    print(i)
```

```python
str="omkar"
for i in str:
    print(i)
```

## While loop

```python
count=0
while(count<5):
    print(count)
    count=count+1
```

```python
count=0
while(count%2==0):
    print(count)
    count=count+1
```

## Break

```python
for i in range(10):
    if(i==5):
        break
    print(i)
```

## continue


```python
for i in range(10):
    if(i%2==0):
        continue
    print(i)
```

## Pass

```python
for i in range(5):
    if(i==3):
        pass
    print(i)
```

## Nested loop

```python
for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

## Examples

```python
## Calculate the sum of first N natural numbers using a while and for loop

#while loop
n=10
sum=0
count=1
while(count<=n):
    sum=sum+count
    count+=1
print(sum)
```

```python
# for loop
sum=0
n=10
for i in range(11):
    sum=sum+i
print(sum)
```

```python
## prime number between 1 and 100

for num in range (1,101):
    if(num>1):
        for i in range(2,num):
            if(num%i==0):
                break

        else:
            print(num)
```
