File handling is a crucial part of any programming language. Python provides built-in functions and methods to read from and write to files, both text and binary

```python
### Read a Whole File

with open('05-example.txt','r') as file:
    content=file.read()
    print(content)
```

```python
## Read a file line by line
with open('05-example.txt','r') as file:
    for line in file:
        print(line)
```

```python
## Read a file line by line
with open('05-example.txt','r') as file:
    for line in file:
        print(line.strip()) ## sstrip() removes the newline character
```

```python

## Writing a file(Overwriting)

with open('05-example.txt','w') as file:
    file.write('Hello World!\n')
    file.write('this is a new line.')
```

```python

## Write a file(without Overwriting)
with open('05-example.txt','a') as file:
    file.write("Append operation taking place!\n")
```

```python

### Writing a list of lines to a file
lines=['First line \n','Second line \n','Third line\n']
with open('05-example.txt','a') as file:
    file.writelines(lines)
```

```python
### Binary Files

# Writing to a binary file
data = b'\x00\x01\x02\x03\x04'
with open('04-example.bin', 'wb') as file: ## write byte
    file.write(data)
```

```python

# Reading a binary file
with open('04-example.bin', 'rb') as file: ## read byte
    content = file.read()
    print(content)
```

```python
### Read the content from  source text fiile and write to a destination text file
# Copying a text file
with open('05-example.txt', 'r') as source_file:
    content = source_file.read()

with open('03-destination.txt', 'w') as destination_file:
    destination_file.write(content)
```

```python
#Read a text file and count the number of lines, words, and characters.
# Counting lines, words, and characters in a text file
def count_text_file(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()
        line_count = len(lines)
        ## "Hello world\n".split() becomes ["Hello", "world"]
        ## same as 
        '''for line in lines:
            words=line.split()
            word_count=word_count+len(words)'''
        word_count = sum(len(line.split()) for line in lines)
         
        char_count = sum(len(line) for line in lines)
    return line_count, word_count, char_count

file_path = '05-example.txt'
lines, words, characters = count_text_file(file_path)
print(f'Lines: {lines}, Words: {words}, Characters: {characters}')
```

```python
# The w+ mode in Python is used to open a file for both reading and writing. If the file does not exist, it will be created. If the file exists, its content is truncated (i.e., the file is overwritten).

### Writing and then reading a file

with open('hello.txt','w+') as file:
    file.write("Hello world\n")
    file.write("This is a new line \n")

    ## Move the file cursor to the beginning else cursor is at last and it wont read ## anything
    file.seek(0)

    ## Read the content of the file
    content=file.read()
    print(content)
```
