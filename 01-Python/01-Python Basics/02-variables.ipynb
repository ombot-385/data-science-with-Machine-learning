```python
## Declaring and assigning variables
#variables name must start with a letter or an '_' and can contains letters, numbers and underscores
##variables names are case sensitive


age=32
height=6.1
name="omkar"

print("age:",age)
print("height:",height)
print("name:",name)
```

```python
## Variables types
## python is dynamically typed , type of a variable is determined at runtime

age=25 #int
height=6.1 #float
name="omkar" #str
is_student=True #boolean

print(type(age))
print(type(height))
print(type(name))
print(type(is_student))
```

```python
##Type conversion
age=25
print(age)
print(type(age))

age_str=str(age)
print(age_str)
print(type(age_str))
```

```python

##Type conversion
height=5.1
print(height)
print(type(height))

heightint=int(height)
print(heightint)
print(type(heightint))

height2=float(int(heightint))
print(height2)
print(type(height2))
```

```python
##Dynamic typing
## Python allows the type of a variable to change as the program executes
var=10 #int
print(var,type(var))

var="hello"
print(var,type(var))

var=True
print(var,type(var))
```

```python
##Input
age=input("what is your age")
print(age,type(age))
```

```python
##simple calculator
num1=float(input("enter your first number"))
num2=float(input("enter your second number"))

sum=num1+num2
difference=num1-num2
product=num1*num2
quotient=num1/num2

print("sum:",sum)
print("difference",difference)
print("product",product)
print("quotient",quotient)

```
