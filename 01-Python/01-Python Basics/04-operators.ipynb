```python
##Arithmatic operations
a=10
b=5

add_res=a+b
sub_res=a-b
mul_res=a*b
div_res=a/b
floor_div_res=a//b
modulus_res=a%b
expo_res=a**b

print(add_res)
print(sub_res)
print(mul_res)
print(div_res)
print(floor_div_res)
print(modulus_res)
print(expo_res)
```

```python
##Comparison operator
a=10
b=10
print(a==b)

str1="omkar"
str2="omkar"
print(str1!=str2)
str1==str2

print("==============================================")
##Greater than,greater than equal to ,less than and less than equal to
num1=45
num2=45
print(num1>num2)
print(num1>=num2)
print(num1<num2)
print(num1<=num2)
```

```python
##Logical operator
##And,Not,OR

x=True
y=False
result=x and y
print(result)

result=x or y
print(result)

z=not y
result=x and z
print(result)
print(not x)
```
